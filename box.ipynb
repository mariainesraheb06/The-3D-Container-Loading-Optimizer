```python
import itertools
from typing import List, Tuple, Optional


class Box:
    def __init__(
        self,
        id: int,
        length: float,
        width: float,
        height: float,
        weight: float = 0.0,
        fragile: bool = False,
    ):
        if length <= 0 or width <= 0 or height <= 0:
            raise ValueError("Box dimensions must be positive")
        if weight <= 0:
            raise ValueError("Weight must be positive")

        self.id = id
        self.length = length
        self.width = width
        self.height = height
        self.weight = weight
        self.fragile = fragile
        self.volume = length * width * height

    def get_orientations(self) -> List[Tuple[float, float, float]]:
        if not self.fragile:
            perms = set(itertools.permutations((self.length, self.width, self.height)))
            return list(perms)
        else:
            orient1 = (self.length, self.width, self.height)
            orient2 = (self.width, self.length, self.height)
            if orient1 == orient2:
                return [orient1]
            return [orient1, orient2]

    def get_base_area(self) -> float:
        return self.length * self.width

    def get_volume(self) -> float:
        return self.volume

    def __repr__(self) -> str:
        return (f"Box(id={self.id}, dims=({self.length:.3f},{self.width:.3f},{self.height:.3f}), "
                f"weight={self.weight:.2f}, fragile={self.fragile})")
```
